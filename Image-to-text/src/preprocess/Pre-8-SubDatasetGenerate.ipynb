```python
import cv2
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
import glob
import json
import os
import sys
import random
import shutil
import h5py

import logging
from logging.handlers import RotatingFileHandler

sys.path.append("../../../../donut/src/test/")
from common.config import cfg
```

```python
log_file = "./logs/SmallSizeDatasetGenerate.log"
logging.basicConfig(
    handlers=[RotatingFileHandler(log_file, maxBytes=10 * 1024 * 1024, backupCount=3)],
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s",
)
```

```python
target_image_dir = "../../../data/sub_data/images/"
target_anno_dir = "../../../data/sub_data/annotations/"
```

```python
# list all image paths
image_dir = "../../../data/image_resize/images/"
image_files = glob.glob(image_dir + "*")
print(len(image_files))
print(image_files[:5])
```

```python
# list all annotation files paths

annotations_dir="../../../data/image_resize/annotations/"
annotations_files=glob.glob(annotations_dir+"*")
print(len(annotations_files))
print(annotations_files[:5])
```

```python
# use 10% data from raw dataset
sample_size = int(len(image_files) * 0.1)

sampled_images = random.sample(image_files, sample_size)

print(f"Total images: {len(image_files)}")
print(f"Sampled images: {sample_size}")
```

```python
for img_path in sampled_images:
    filename = os.path.basename(img_path)
    basename = os.path.splitext(filename)[0]

    # corresponding annotation files
    json_path = os.path.join(annotations_dir, basename + ".json")

    # target destination
    shutil.copy(img_path, os.path.join(target_image_dir, filename))
    if os.path.exists(json_path):
        shutil.copy(json_path, os.path.join(target_anno_dir, basename + ".json"))
    else:
        logging.error(f"Missing-Annotation not found for {basename}")

print("Sampling complete.")
```
