```python
# list all annotation files paths

annotations_dir="../../../data/image_resize/annotations/"
annotations_files=glob.glob(annotations_dir+"*")
print(len(annotations_files))
print(annotations_files[:5])
```

```python
def categorize_json_by_chart_type(json_paths, output_dir="/Users/yiding/personal_projects/ML/github_repo/donut/data/image_classes"):
    """
    Give a array of json paths, and write same type of json files into .txt file

    params:
        json_paths (list[str]): all paths to json files
        output_dir (str): output .txt file
    """
    valid_types = {"dot", "scatter", "horizontal_bar", "line", "vertical_bar"}

    os.makedirs(output_dir, exist_ok=True)

    categorized = defaultdict(list)

    for json_path in json_paths:
        try:
            with open(json_path, "r") as f:
                data = json.load(f)
            chart_type = data.get("chart-type", "").lower()
            if chart_type in valid_types:
                file_name = os.path.basename(json_path)
                categorized[chart_type].append(file_name)
                # print(f"{file_name} → {chart_type}.txt")
            else:
                logging.error(f"UNKNOWN chart-type: {chart_type}")
        except Exception as e:
            logging.error(f"Missing process: {json_path}")

    for chart_type, files in categorized.items():
        txt_path = os.path.join(output_dir, f"{chart_type}.txt")
        with open(txt_path, "w") as out_f:
            for f in files:
                out_f.write(f + "\n")
```

```python
categorize_json_by_chart_type(annotations_files)
```

```python
def image_classify(raw_path_prefix:str,file_txt_path:str,dst_folder:str):
    """
    Move images with the same types into folder
    """

    os.makedirs(dst_folder, exist_ok=True)

    with open(file_txt_path, "r") as f:
        lines = [line.strip() for line in f if line.strip()]

    for json_name in lines:
        stem = os.path.splitext(json_name)[0]
        src_img_path = os.path.join(raw_path_prefix, f"{stem}.jpg")
        dst_img_path = os.path.join(dst_folder, f"{stem}.jpg")

        if os.path.exists(src_img_path):
            shutil.copy(src_img_path, dst_img_path)
            logging.info(f"Moved: {src_img_path} → {dst_img_path}")
        else:
            logging.warning(f"Missing image: {src_img_path}")

    return
    
```

```python
# vertical_bar images

vertical_bar_img_path_prefix = (
    "/Users/yiding/personal_projects/ML/github_repo/donut/data/image_resize/images/"
)
vertical_bar_dst_folder = "/Users/yiding/personal_projects/ML/github_repo/donut/data/image_classes/vertical_bar"


image_classify(
    raw_path_prefix=vertical_bar_img_path_prefix,
    file_txt_path="/Users/yiding/personal_projects/ML/github_repo/donut/data/image_classes/vertical_bar.txt",
    dst_folder=vertical_bar_dst_folder,
)
```
